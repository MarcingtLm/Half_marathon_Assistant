```python
import pandas as pd

df_2023 = pd.read_csv('cleaned_halfmarathon_wroclaw_2023.csv', sep=';')
df_2024 = pd.read_csv('cleaned_halfmarathon_wroclaw_2024.csv', sep=';')

print(f"Kolumny 2023: {df_2023.columns.tolist()}")
print(f"Kolumny 2024: {df_2024.columns.tolist()}")
print(f"\nLiczba wierszy 2023: {len(df_2023)}")
print(f"Liczba wierszy 2024: {len(df_2024)}")
```

```python
df_2024 = df_2024.drop(columns=['Rocznik'])

df_2023['Rok'] = 2023
df_2024['Rok'] = 2024

print(f"Kolumny 2023 po modyfikacji: {df_2023.columns.tolist()}")
print(f"Kolumny 2024 po modyfikacji: {df_2024.columns.tolist()}")
```

```python
df_combined = pd.concat([df_2023, df_2024], ignore_index=True)
print(f"Liczba wierszy po połączeniu: {len(df_combined)}")
print(f"Kolumny: {df_combined.columns.tolist()}")
df_combined.head()
```

```python
df_combined.to_csv('combined_halfmarathon_wroclaw.csv', sep=';', index=False)
```

```python
df = df_combined.copy()
```

```python
df.columns
```

```python
from pycaret.regression import setup, compare_models, predict_model, models


exp = setup(
    data=df,
    target='Czas',  
    categorical_features=['Płeć'],
    numeric_features=['Wiek', '5 km Czas'],
    ignore_features=[
        '10 km Czas', '10 km Miejsce Open', '10 km Tempo',
        '15 km Czas', '15 km Miejsce Open', '15 km Tempo',
        '20 km Czas', '20 km Miejsce Open', '20 km Tempo',
        'Miejsce', 'Płeć Miejsce', 'Kategoria wiekowa Miejsce', 
        'Tempo', 'Tempo Stabilność', '5 km Miejsce Open', '5 km Tempo',
         'Kategoria wiekowa','Rok'
    ],   
    # transformation=True,
    # transformation_method='yeo-johnson',
    normalize=True,
    normalize_method='zscore',
    session_id=123,
    verbose=False
)
```

```python
best_models = exp.compare_models(sort='R2',n_select=3)
tuned_best_models = [exp.tune_model(m, optimize="MAE") for m in best_models]
best_model = exp.compare_models(best_models + tuned_best_models, sort="R2")
```

```python
exp.predict_model(best_model)
```

```python
exp.plot_model(best_model, plot="residuals")
```

```python
exp.plot_model(best_model, plot="error")
```
